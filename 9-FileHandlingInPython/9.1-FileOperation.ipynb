```python
## Read Whole File

with open('example.txt','r') as file:
    content = file.read()
    print(content)
```

```python
## Read file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() remove newline character
```

```python
## Write in a file(overwritten)

with open('example.txt','w') as file:
    file.write("Hello World!\n")
    file.write("This is New Line.\n")
```

```python
## Writing in a file without Overwriting

with open('example.txt','a') as file:
    file.write("This is written by using append mode.\n")
```

```python
## Writing a list of lines to a file
lines = ['First Line \n','Second Line \n','Third Line \n']

with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## Binary Files

data = b'\x00\x01\x02\x03\x04'

with open('example.bin','wb') as file:
    file.write(data)
```

```python
## Read Binary File

with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
## Read content from sorce text file and write in destination text file
with open('example.txt','r') as file:
    content = file.read()

with open('destination.txt','w') as file:
    file.write(content)    
```

```python
## Read a text file and count number of lines,words and characters.
def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count,word_count,char_count

file_path = 'example.txt'
lines,words,characters = count_text_file(file_path)
print(f"Lines : {lines}, Words : {words}, Characters : {characters}")    
    
```

```python
## Read and write in text file

with open('example.txt','w+') as file:
    file.write("Hello World!\n")
    file.write("This is newline\n")

    file.seek(0)

    content = file.read()
    print(content)
```
