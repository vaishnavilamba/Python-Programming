```python
import pandas as pd
```

```python
data = [1,2,3,4,5,6]
series = pd.Series(data)
print("Series : \n",series)
print(type(series))
```

```python
# Create series from dictionary
dict = {'a':1,'b':2,'c':3,'d':4}
series = pd.Series(dict)
print(series)
```

```python
data = [10,20,30]
index = ['a','b','c']
series = pd.Series(data,index = index)
print(series)
```

```python
data = {
    'Name':["Monica","Ross","Rach"],
    'City':["Banglore","NewYork","America"],
    'Age':[28,26,28]
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
# Create a dataFame from list of Dictionaries

data = [
    {'Name':"Monica",'City':"America",'Age':28},
    {'Name':"Ross",'City':"New York",'Age':30},
    {'Name':"Rach",'City':"Germany",'Age':28},
    {'Name':"Phebe",'City':"London",'Age':31},
    {'Name':"Joey",'City':"India",'Age':29},
    {'Name':"Chandler",'City':"America",'Age':28}
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
