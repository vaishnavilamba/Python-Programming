```python
## Data Serialization

import json
data = {'Name':'Vaishnavi','Age':20,'CGPA':9.3}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
##CSV

import csv

with open('source.txt',mode = 'w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Name","Age"])
    writer.writerow(["Vaishnavi",20])

with open('source.txt',mode = 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)    
```
