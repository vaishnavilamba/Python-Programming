```python
import sqlite3 as sq
```

```python
connection = sq.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS employees(
    id Integer Primary key,
    name Text NOT NULL,
    age Integer,
    department Text
)
''')
```

```python
connection.commit()
```

```python
cursor.execute('SELECT * FROM employees')
```

```python
## Insert the data in sqlite table 
cursor.execute(''' 
INSERT INTO employees(name,age,department)
VALUES('Bob',25,'Engineering')
''')

cursor.execute(''' 
INSERT INTO employees(name,age,department)
VALUES('Charlie',26,'Finance')
''')

cursor.execute(''' 
INSERT INTO employees(name,age,department)
VALUES('Joey',28,'IT')
''')

connection.commit()
```

```python
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()
```

```python
for row in rows:
    print(row)
```

```python
cursor.execute('''
UPDATE employees
SET age = 34
WHERE name = 'Bob'
''')
connection.commit()
```

```python
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute('''
DELETE from employees
WHERE name = 'Bob'
''')
connection.commit()
```

```python
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
## Working with sales Data
connection = sq.connect('Sales_Data.db')
cursor = connection.cursor()
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
   id Integer Primary key,
   date TEXT NOT NULL,
   product TEXT NOT NULL,
   sales Integer,
   region Text
)
''')

connection.commit()
```

```python
sales_data = [
    ('2023-01-01','Product1',100,'North'),
    ('2023-01-02','Product2',200,'South'),
    ('2023-01-03','Product3',150,'East'),
    ('2023-01-04','Product4',250,'West'),
    ('2023-01-05','Product5',300,'North')
]
```

```python
cursor.executemany('''
INSERT INTO sales(date,product,sales,region)
VALUES(?,?,?,?)
''',sales_data)
connection.commit()
```

```python
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
connection.close()
```
