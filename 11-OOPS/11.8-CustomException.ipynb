```python
class Error(Exception):
    pass

class dobError(Error):
    pass
```

```python
year = int(input("Enter your age.."))
age = 2026-year
print(f"Entered Age : {age}")
try:
    if(age<=30 and age>=20):
        print("The age is valid,you can apply for an exams.")
    else:
        raise dobError
except dobError:
    print("Sorry,your age should be greater than 20 or smaller than 30.") 
```
