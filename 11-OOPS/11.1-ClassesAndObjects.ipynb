```python
## Instance Variable and Methods

class Dog:
    ##Constructor 
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woofs")    

#Create and object
panthu = Dog("Panther",4)
print(panthu)
print(panthu.name)
print(panthu.age)  
panthu.bark()      
```

```python
## Modeling a Bank Account

# Create an class
class BankAccount:

    ## Create a constructor
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"{amount} is deposited. New Balanace is {self.balance}")    \

    def withdraw(self,amount):
        if amount>self.balance:
            print(f"Insufficent Balance")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawed. New Balance is {self.balance}")


    def get_balance(self):
        return self.balance                
  
```

```python
## creating an object of BankAccount class

user1 = BankAccount("Vaishnavi")
```

```python
user1.deposit(10000)
```

```python
user1.withdraw(500)
```

```python
user1.get_balance()
```
