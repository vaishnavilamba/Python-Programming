```python
## Demonstrate Polymorphism

#Parent Class
class Animal:
    def __init__(self,name):
        self.name = name

    def speak(self):
        return "Animal can speak!" 

class Dog(Animal):
    def __init__(self,name):
        super().__init__(name)       

    def speak(self):
        return "Woof!"

class Cat(Animal):
    def __init__(self,name):
        super().__init__(name)       
    
    def speak(self):
        return "Meow!"

def animal_speak(animal):
    print(animal.speak())    


animal = Animal("Bear")
Dog = Dog("Panther")
Cat = Cat("Kitti")

animal_speak(animal)
animal_speak(Dog)
animal_speak(Cat)

```

```python
class Shape:
    def area(self):
        return "The area of Figure"

class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width*self.height

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return 3.14*self.radius*self.radius

def shape_area(shape):
    print(f"The area {shape.area()}")

circle = Circle(3)
rectangle =  Rectangle(3,4)                     
shape = Shape()

shape_area(circle)
shape_area(rectangle)
shape_area(shape)
```
