```python
class Car:

    def __init__(self,enginetype,windows,doors):
        self.engineType = enginetype
        self.windows = windows
        self.doors = doors

    def drive(self):
        print(f"The person drive {self.engineType} car")


tata = Car("Petrol",4,4)
print(tata)
tata.drive()            
```

```python
class Tesla(Car):

    def __init__(self,windows,doors,enginetype,is_selfDriving):
        super().__init__(enginetype,windows,doors)
        self.is_selfDriving = is_selfDriving

    def selfDriving(self):
        print(f"Tesla Support self Driving : {self.is_selfDriving}")    
```

```python
tesla = Tesla(4,4,"Electric",True)
print(tesla)
tesla.selfDriving()
tesla.drive()
```

```python
## Multiple Inheritance

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclass must implement this method")
```

```python
class Pet:
    def __init__(self, owner):
        self.owner = owner

    def get_owner(self):
        return self.owner  
```

```python
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def doginfo(self):
        print(f"Owner name is {self.owner} and Pet is {self.name}")  
```

```python
panthu = Dog("Panther", "Vaishnavi")

panthu.speak()
print(panthu.get_owner())
panthu.doginfo()
```
