```python
#Nested Dictionaries
family = {
    "Mother" : {"name":"Anuradha","Age":42},
    "Father" : {"name":"Manish","Age":52},
    "Brother": {"name":"Aman","Age":17},
    "Sister" : {"name":"Vaishnavi","Age":21},
    "Pet"    : {"name":"Panther","Age":16}
}
family
```

```python
#Access Nested Dictionaries
print(family["Pet"]["name"])
print(family["Pet"]["Age"])
```

```python
##Iterating Over Nested Dictionary
for family_realtion,family_info in family.items():
    print(f"{family_realtion} -> ")
    for key,value in family_info.items():
        print(f"{key}:{value}")
```

```python
#Dicationary Comperhension

squares = {x:x**2 for x in range(11)}
squares
```

```python
even_squares = {x:x**2 for x in range(11) if x%2==0 }
even_squares
```

```python
numbers = [1,2,3,4,2,4,2,4,2,1,8,5,7,9]
freq = {}

for number in numbers:
    if number in freq:
        freq[number] += 1
    else:
        freq[number] = 1    

print(freq)
```
