```python
## Exception try , except block

try:
    a = b
except:
    print("The variable has not been assign")    
```

```python
##NameError → the type of exception being caught.
##ex → a variable that contains the details of that exception.

try:
    a = b
except NameError as ex:
    print(ex)    
```

```python
num=10
print(num/0)
```

```python
try:
    num/0
except:
    print("The number cannot be divided by zero.")    
```

```python
try:
    num/0
except ZeroDivisionError as e:
    print(f"Error occured, {e}")    
```

```python
## Using parent class exception

try:
    num/2
    a=b
except ZeroDivisionError as e:
    print(f"Error occured, {e}")    
except Exception as e1:
    print(e1)    
    print("Main exception got caught here")
```

```python
#try,except,else and finally

try:
    num = int(input("Enter number : "))
    result = 10/num
except ValueError:
    print("Enter valid number")
except ZeroDivisionError:
    print("We can't divide by zero")    
except Exception:
    print(ex)    
else:
    print(result)  
finally:
    print("The Excecution completed") # It will always run wether 
                                      # the error is comming or not.
```

```python
## File Handling and Exception Handling
try:
    file=open('example.txt','r')
except Exception as e:
    print(e) 
else:
    # content = file.read()
    # print(content)
    for line in file:
        print(line.strip())
finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print("Successfully file closed")        
```
