```python
#elif Statement
day = 7
if(day==1):
    print("Monday")
elif(day==2):
    print("Tuesday")
elif(day==3):
    print("Wednesday")
elif(day==4):
    print("Thursday")
elif(day==5):
    print("Friday")
elif(day==6):
    print("Saturday")
else:
    print("Sunday")                    
```

```python
#Nested Condition Statement
num = -67
if(num>0):
    print("Num is positive")
    if(num%2==0):
        print("Even")
    else:
        print("Odd")
else:
    print("The zero or Negative number")     
    if(num%2==0):
        print("Even")
    else:
        print("Odd")   
```

```python
##Use cases or Practical examples
#Simple Calculator
num1 = float(input("Enter your first number : "))
num2 = float(input("Enter your second number"))
op = input("Enter operand")
if(op=='+'):
    print(num1+num2)
elif(op=='-'):
    print(num1-num2)
elif(op=='*'):
    print(num1*num2)
elif(op=='/'):
    print(num1/num2)
elif(op=="%"):
    print(num1%num2)    
else:
    print("Invalid operand")    
                
```

```python
#Break
for i in range(10):
    if(i==5):
        break
    print(i)
```

```python
#Continue
for i in range(10):
    if(i==5):
        continue
    print(i)
```

```python
#Nested Loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
#Sum N Natural Numbers using for Loop
sum = 0
n = int(input("Enter number : "))
for i in range(n+1):
    sum +=i

print(sum)
```

```python
#Sum N Natural Numbers using while Loop
sum = 0
count = 1
n = int(input("Enter a number : "))
while count<=n:
    sum+=count
    count = count+1

print(sum)
```
