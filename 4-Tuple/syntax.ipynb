```python
## Packing and Unpacking tuple
## Packing
packed_tuple = 1,"Hello World",3.14
packed_tuple
```

```python
##Unpack Tuple 
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with star
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested Tuple

lst = [[9,8,7,6],["Hi",78],[100,"Aman",4.5]]
lst[0][3:6]
```

```python
tp = (("Hii",99,9.3,True),(1,2,3,4,5,6),('a','b','c'))

for sub_tuple in tp:
    for item in sub_tuple:
        print(item,end=" ")
    print()    
```
