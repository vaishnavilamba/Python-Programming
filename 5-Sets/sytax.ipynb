```python
## Set Memebership Test
my_set = {1,2,3,4,5,6}
print(my_set)
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical Operation
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}


##Union
union_set = set1.union(set2)
print(union_set)

##Intersection
inter_set = set1.intersection(set2)
print(inter_set)


set1.intersection_update(set2)
print(set1)

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

##Difference
print(set1.difference(set2))

## Summetric Differnce
print(set1.symmetric_difference(set2))
```

```python
## Set Methods
set1 = {1,2,3}
set2 = {3,4,5}

# is subset
print(set1.issubset(set2))

print(set1.issuperset(set2))
```
