Creating a TODO List to Keep Track Of Tasks

```python
## Creating todo list
todo_list = ["Buy Fruits","Visit Mandir","Asignments"]

## Adding items to todo list
todo_list.append("Go to Runing")
todo_list.append("Schedule Classes")
todo_list.append("Pay fees")

if "Pay fees" in todo_list:
    print("Need to Pay fee urgently...")

print("ToDo List Remaning")    

for task in todo_list:
    print(f"-->{task}")
```

Create a list of Grades and Calculate average Grade

```python
## List og Grades

grades = [90,89,99,97,50,44]

## Adding grade
grades.append(100)

average_grade = sum(grades)/len(grades)

print(f"Average Grade: {average_grade}")

## Find highest and lowest Grade

highest_grade = max(grades)
lowest_grade = min(grades)

print(f"Lowest Grade -> {lowest_grade}")
print(f"Highest Grade -> {highest_grade}")
```

Use a List to analyze and collect user feedback

```python
feedback = ["Great Service","Very Satisfied!","Cold be better","Excellent experience"]

## Adding new feedabck
feedback.append("Not happy with service")

## Counting Satisfy Feedback 
positive_feedback = sum(1 for comment in feedback if "better" in comment.lower() or "excellent" in comment.lower())
print(f"Positive Feedback Count : {positive_feedback}")

print("User Feedback")

for comment in feedback:
    print(f"-->{comment}")
```
