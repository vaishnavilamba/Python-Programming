```python
lst = [1,2,3,4,5,6,7,8,9]
print(lst[2:5]) #3,4,5
print(lst[:5])  #1,2,3,4,5
print(lst[5:])  #6,7,8,9
print(lst[::2])  #1,3,5,7,9
print(lst[::-1]) #9,8,7,6,5,4,3,2,1
```

```python
# List Comperhension with conditions
even_numbers = [num for num in range(10) if(num%2==0)]
even_numbers
```

```python
#Nested List Comperhension
lst1 = [1,2,3,4,5,6]
lst2 = [-1,-2,-3,-4,-5,-6]

pair = [[i,j] for i in lst1 for j in lst2]

pair
```
