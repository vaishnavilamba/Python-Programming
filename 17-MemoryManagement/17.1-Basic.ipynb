```python
import gc
class MyObject:
    def __init__(self,name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")

obj1 = MyObject("Obj1")
obj2 = MyObject("Obj2")

obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

gc.collect()
```

```python
def generate_number(n):
    for i in range(n):
        yield i
        
```

```python
for num in generate_number(10000):
    print(num)
    if(num>10):
        break
```

```python
import tracemalloc
```

```python
def create_list():
    return [i for i in range(10000)]
```

```python
def main():
    tracemalloc.start()
    create_list()
    snapshot = tracemalloc.take_snapshot()
    top_starts = snapshot.statistics("lineno")
    print("[Top 10]")
    for stat in top_starts[:10]:
        print(stat)
```

```python
main()
```
