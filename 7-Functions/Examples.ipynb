Q:1 Temperature Conversion

```python
def tempConv(temp,unit):
    """This Function convert unit of temperature in Celsius or Fehrenheit"""
    if unit=='C':
        return temp * 9/5 + 32
    elif unit=='F':
        return (temp-32)*5/9
    return None

print(tempConv(45,'C'))
print(tempConv(113,'F'))
```

Q:2 Check the Password is Strong ot not

```python
def is_strong_passwrod(password):
    """This Function check whether the passwrod is strong or not"""
    if len(password)<8:
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char in '!@#$%^&*()+_-' for char in password):
        return False
    return True

print(is_strong_passwrod("123abcD^"))
```

Q:3 Calculate the total Cost of Items in the Cart of Items in the Shooping Cart

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item["price"]*item["quantity"]

    return total_cost


cart = [
    {"name":"Banana","price":0.5,"quantity":6},
    {"name":"Apple","price":8.9,"quantity":8},
    {"name":"Mango","price":7.0,"quantity":7}
]

print(calculate_total_cost(cart))
```

Q:4 Check if a String is Palindrome or not

```python
def is_palindrom_not(s):
    s = s.lower().replace(" ","")
    reversed = s[::-1]
    if s == reversed:
        return True

    return False

print(is_palindrom_not("madam"))
```

Q:5 Factorial of number using Recursion

```python
def fact(num):
    if num == 1 or num == 0:
        return 1
    return num * fact(num-1)

print(fact(5))
```

Q:6 A function to Read A file and count the frequency of each word 

```python
def count_word_freq(filepath):
    word_frq = {}
    with open(filepath,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('.,/:?\"''')
                word_frq[word] = word_frq.get(word,0)+1

    return word_frq

filepath = 'sample.txt'
print(count_word_freq(filepath))
```
