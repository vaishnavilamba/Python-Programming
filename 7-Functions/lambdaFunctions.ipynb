```python
## Syntax

lambda arguments : expression
```

```python
addition = lambda a,b : (a+b)
print(type(addition))
print(addition(3,2))
```

```python
even_or_not = lambda num : num%2==0

print(even_or_not(90))
```

```python
numbers = [1,2,3,4,5]
```

```python
list(map(lambda x:x**2,numbers))
```
