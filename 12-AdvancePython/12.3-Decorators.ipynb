```python
# Closures -> Function inside Function

def main_welcome(topic):
    #msg = "Closure"
    def sub_welcome_method():
        print("Welcome to the advance course of Python.")
        print(topic)
        print("Please learn this concept properly.")
    return sub_welcome_method()    
```

```python
main_welcome("Closure")
```

```python
def main_welcome(func):
    #msg = "Closure"
    def sub_welcome_method():
        print("Welcome to the advance course of Python.")
        func("Closure")
        print("Please learn this concept properly.")
    return sub_welcome_method()  
```

```python
main_welcome(print)
```

```python
def main_welcome(func,lst):
    #msg = "Closure"
    def sub_welcome_method():
        print("Welcome to the advance course of Python.")
        print(func(lst))
        print("Please learn this concept properly.")
    return sub_welcome_method()  
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

```python
## Decorator
def main_welcome(func):
    #msg = "Closure"
    def sub_welcome_method():
        print("Welcome to the advance course of Python.")
        func()
        print("Please learn this concept properly.")
    return sub_welcome_method()    
```

```python
def course_intro():
    print("This is Machine Larning Course and advance Python.")
    
course_intro()
       
```

```python
main_welcome(course_intro)
```

```python
@main_welcome
def course_intro():
    print("This is Machine Larning Course and advance Python.")
```

```python
def intro(func):
    def wrapper():
        print("Hii My name is vaishu.")
        func()
        print("I am 21 year old.")

    return wrapper()    
```

```python
@intro
def nickname():
    print("My nickname is Vaishu")
```
