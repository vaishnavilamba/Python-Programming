```python
def square(n):
    for i in range(n):
        return i**2    
```

```python
square(3)
```

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
def generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = generator()
gen
```

```python
print(next(gen))
```

```python
gen = generator()
```

```python
try:
    print(next(gen))
except StopIteration: 
    print()   
```

```python
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
## Reading Large Files
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
